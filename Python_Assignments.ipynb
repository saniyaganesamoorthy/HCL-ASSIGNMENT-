1. Print all prime numbers between input range

```python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

print("Prime numbers:")

for num in range(start, end + 1):
    if num > 1:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

2. Factorial using recursion

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input("Enter a number: "))

print("Factorial:", factorial(n))
```

3. Square of numbers using lambda

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print("Numbers:", numbers)
print("Squares:", squares)
```

5. Count frequency of characters in a string

```python
text = input("Enter a string: ")

frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print("Character frequency:")

for char, count in frequency.items():
    print(char, ":", count)
```

6. Calculate area of a circle using math library

```python
import math

radius = float(input("Enter radius: "))

area = math.pi * radius * radius

print("Area of circle:", area)
```

7. Reverse a string without using built-in reverse

```python
text = input("Enter a string: ")

reverse = ""

for char in text:
    reverse = char + reverse

print("Reversed string:", reverse)
```

8. Remove duplicates from a list

```python
numbers = [10, 20, 10, 30, 20, 40, 30]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("Original list:", numbers)
print("List after removing duplicates:", unique)
```

9. Merge two dictionaries

```python
dict1 = {"name": "Saniya", "age": 20}
dict2 = {"course": "AI & ML", "college": "Saveetha"}

merged = dict1.copy()
merged.update(dict2)

print("Merged dictionary:", merged)
```

10. Fibonacci series using recursion

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter number of terms: "))

print("Fibonacci series:")

for i in range(n):
    print(fibonacci(i), end=" ")
```
