```python
numbers = input().split(',')
result = []
for num in numbers:
    if int(num, 2) % 5 == 0:
        result.append(num)
print(','.join(result))
```

```python
sentence = input()
letters = 0
digits = 0
for ch in sentence:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
print("LETTERS", letters)
print("DIGITS", digits)
```

```python
n = int(input())
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(factorial)
```
